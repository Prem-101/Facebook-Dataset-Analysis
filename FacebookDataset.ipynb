#### **Question 1: How does the time of upload (`status_published`) affect the `num_reactions`?**

To understand the impact of upload time on reactions, we extract the hour of the day, day of the week and month from the `status_published` column. Then, we analyse the average number of reactions for different hours and days.

```python
df['hour']=df['status_published'].dt.hour
df['day_of_week']=df['status_published'].dt.day_name()
df['month']=df['status_published'].dt.month_name()

df
```

```python
from matplotlib import pyplot as plt
import seaborn as sns
df.groupby('day_of_week').size().plot(kind='barh', color=sns.palettes.mpl_palette('Dark2'))
plt.gca().spines[['top', 'right',]].set_visible(False)
```

```python
hourly_reactions=df.groupby('hour')['num_reactions'].mean().reset_index()
daily_reactions=df.groupby('day_of_week')['num_reactions'].mean().reindex(['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday'])
```

```python
monthly_reactions = df.groupby('month')['num_reactions'].mean().reindex(['January', 'February', 'March', 'April', 'May', 'June',
                                'July', 'August', 'September', 'October', 'November', 'December'])
```

```python
from matplotlib import pyplot as plt
import seaborn as sns
def _plot_series(series, series_name, series_index=0):
  palette = list(sns.palettes.mpl_palette('Dark2'))
  xs = series['hour']
  ys = series['num_reactions']

  plt.plot(xs, ys, label=series_name, color=palette[series_index % len(palette)])

fig, ax = plt.subplots(figsize=(10, 5.2), layout='constrained')
df_sorted = hourly_reactions.sort_values('hour', ascending=True)
_plot_series(df_sorted, '')
sns.despine(fig=fig, ax=ax)
plt.xlabel('hour')
_ = plt.ylabel('num_reactions')
```

```python
plt.figure()
sns.barplot(x=daily_reactions.index, y=daily_reactions.values)
plt.title('Average Number of Reactions by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Average Reactions')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
monthly_reactions_df = monthly_reactions.reset_index()

sns.barplot(data=monthly_reactions_df, x='month', y='num_reactions')
plt.title('Average Reactions by Month')
plt.xlabel('Month')
plt.ylabel('Average Reactions')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

##### **Question 3:	Use the columns `status_type`, `num_reactions`, `num_comments`, `num_shares`, `num_likes`, `num_loves`, `num_wows`, `num_hahas`, `num_sads`, and `num_angrys` to train a K-Means clustering model on the Facebook Live Sellers dataset.***

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.cluster import KMeans
from sklearn.compose import ColumnTransformer

data=df[['status_type', 'num_reactions', 'num_comments', 'num_shares', 'num_likes', 'num_loves', 'num_wows', 'num_hahas', 'num_sads', 'num_angrys']].copy()
```

```python
data
```

```python
preprocessor=ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop='first', sparse_output=False), ['status_type']),
        ('num', StandardScaler(), ['num_reactions', 'num_comments', 'num_shares', 'num_likes', 'num_loves', 'num_wows', 'num_hahas', 'num_sads', 'num_angrys'])
    ]
)
X_preprocessed=preprocessor.fit_transform(data)
```

```python
X_preprocessed
```

```python
kmeans=KMeans(n_clusters=3, random_state=42)
data.loc[:, 'cluster']=kmeans.fit_predict(X_preprocessed)
```

```python
data
```

```python
cluster_summary=data.groupby('cluster').mean(numeric_only=True)
cluster_types=data.groupby('cluster')['status_type'].value_counts().unstack().fillna(0)

print("Cluster Summary (Mean Values):\n", cluster_summary)
print("\nCluster Status Types:\n", cluster_types)
```

##### **Question 4:	Use the elbow method to find the optimum number of clusters.**

```python
inertias=[]
k_range=range(1, 20)

for k in k_range:
    kmeans=KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_preprocessed)
    inertias.append(kmeans.inertia_)
```

```python
plt.figure()
plt.plot(k_range, inertias, marker='o')
plt.title('Elbow Method for Optimal Number of Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Intertia')
plt.grid(True)
plt.show()
```
